```python
import matplotlib.pyplot as plt

Months=  ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct"]
Valuse = [10,50,90,70,97,45,66,80,72,93]

bars= plt.bar(Months,Valuse,color="skyblue", edgecolor="Black")

plt.bar_label(bars,padding=3)
plt.ylim(0,120)

plt.title("Monthly_Data_Bar_Chart")
plt.xlabel("Months")
plt.ylabel("Values")

plt.show()
```
